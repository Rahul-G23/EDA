```python
#12.	Create a 3x4 subarray view from a larger array, modify one of its elements, and print the original array to demonstrate that slicing returns a view, not a copy. Then repeat using .copy() and show that the original is unaffected.
x = np.random.randint(0, 10, (5, 6))
subarray = x[:3, :4]
subarray[0, 0] = 99
print("Original array after modifying subarray:")
print(x)

# Using .copy()
subarray_copy = x[:3, :4].copy()
subarray_copy[0, 0] = 88
print("Original array after modifying subarray copy:")
print(x)
```

```python
#12.	Create a 3x4 subarray view from a larger array, modify one of its elements, and print the original array to demonstrate that slicing returns a view, not a copy. Then repeat using .copy() and show that the original is unaffected.
y = np.arange(12).reshape(4, 3)
subarray = y[1:4, 0:3]
print("Original array:")
print(y)
subarray[0, 0] = 99
print("Modified subarray:")
print(subarray)
print("Original array after modifying subarray:")
print(y)

y = np.arange(12).reshape(4, 3)
subarray = y[1:4, 0:3].copy()
print("Original array:")
print(y)
subarray[0, 0] = 99
print("Modified subarray:")
print(subarray)
print("Original array after modifying subarray:")
print(y)
```

```python
#13.	Create a 1-D array of 12 elements and reshape it into a 3x4 matrix. Then reshape the same original array into a column vector using both reshape((12,1)) and x[:, np.newaxis], and confirm the two results are identical.
x = np.arange(12)
y = x.reshape(3, 4)
z1 = x.reshape((12, 1))
z2 = x[:, np.newaxis]
print("Original array:")
print(x)
print("Reshaped into 3x4 matrix:")
print(y)
print("Column vector using reshape((12, 1)):")
print(z1)
print("Column vector using x[:, np.newaxis]:")
print(z2)
print("Are the two column vectors identical?")
print(np.array_equal(z1, z2))
```
