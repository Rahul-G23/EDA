```python
import numpy as np
import pandas as pd
```

```python
data=pd.Series([1,np.nan,'hello',None])
print(data)
print()
print(data.isnull())
print()
print(data.dropna())
```

```python
df=pd.DataFrame([[1,np.nan,2,5,9],[np.nan,54,8,3,7],[7,6,np.nan,78,3],[26,46,79,35,np.nan],[45,35,86,np.nan,7]])
print(df)
print()
print(df.isnull())
print()
print(df.sum())
```

```python
print(df.dropna())
print()
print(df.dropna(axis='columns'))
```

```python
df1=pd.DataFrame([[1,np.nan,2,5,9],[np.nan,54,8,3,7],[np.nan,np.nan,np.nan,np.nan,np.nan],[26,46,79,35,np.nan],[45,35,86,np.nan,7]])
print(df1)
print()
print(df1.dropna(how='all'))
```

```python
df3=pd.Series([1, np.nan, 2, None, 3, np.nan])
print(df3)
print()
print(df3.fillna(0))
print()
print(df3.ffill())
print()
print(df3.bfill())
```

```python
#Create a small DataFrame simulating 7 days of temperature readings with 2 missing days. Fill the missing values using the column mean, and separately using linear interpolation via df.interpolate().
df4=pd.DataFrame({'Day':['Mon','Tue','Wed','Thu','Fri','Sat','Sun'],'Temperature':[22,24,np.nan,26,28,np.nan,30]})
print(df4)
print()
# Fill missing values with column mean
df4['Temperature'].fillna(df4['Temperature'].mean(), inplace=True)
print(df4)
```
