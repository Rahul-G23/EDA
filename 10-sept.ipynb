```python
import numpy as np
```

```python
arr= np.array([10,20,30,40,50])
print(arr)
print(type(arr))
```

```python
ar= np.array([[1,2],[3,4],[5,6]])
print(ar)
```

```python
x= np.array([7,8,9],dtype=float)
print(x , type(x))
```

```python
y=np.zeros(6)
print(y)
```

```python
z=np.ones((3,2),dtype=int)
print(z)
```

```python
a=np.empty((2,3))
print(a)
'''why we use np.empty() over np.zeros()
because This can be useful when we want to create an array and fill it with values later, 
without the overhead of initializing it to zero first.'''
```

```python
b=np.arange(0,20,2)
print(b)
```

```python
c=np.linspace(0,1,6)
print(c)
'''in arange they will exclude the last number but in linspace they will include the last number'''
```

```python
d=np.random.random((3,3))
print(d)
```
