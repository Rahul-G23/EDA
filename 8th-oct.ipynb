```python
import numpy as np
import pandas as pd
```

```python
index=[['A',2010],['A',2020],['A',2025],['B',2010],['B',2020],['B',2025]]
population=[123,456,789,102,203,304]
i=pd.MultiIndex.from_tuples(index,names=['Country','Year'])
print(i)
pop=pd.Series(population,index=i)
print(pop)
```

```python
index=[['A',2010],['A',2020],['A',2025],['B',2010],['B',2020],['B',2025]]
population=[1.5,1.7,1.9,2.5,2.7,2.9]
i=pd.MultiIndex.from_product(index)
print(i)
```

```python
index=[['A',2010],['A',2020],['A',2025],['B',2010],['B',2020],['B',2025]]
population=[123,456,789,102,203,304]
i1=pd.MultiIndex.from_arrays([['A','A','A','B','B','B'],[2010,2020,2025,2010,2020,2025]],names=['Country','Year'])
print(i1)
pop=pd.Series(population,index=i1)
print(pop)
```

```python
i2=pd.MultiIndex.from_product([['A','B'],[2010,2020,2025]],names=['Country','Year'])
print(i2)
```

```python
l3=pd.MultiIndex(levels=[['A','B'],[2010,2020,2025]],codes=[[0,0,0,1,1,1],[0,1,2,0,1,2]],names=['Country','Year'])
print(l3)
```
