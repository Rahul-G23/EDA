```python
import numpy as np
```

```python
name=['Alice','Bob','Charlie','David']
age=[24,45,37,19]
weight=[55.0,85.3,72.1,68.4]
```

```python
data=np.zeros(
    4,
        dtype={
        'names':('name','age','weight'),
        'formats':('U10','i4','f8')
    }
)
```

```python
print("Structured array using a compound data type")
print(data.dtype)
```

```python
data['name']=name
data['age']=age
data['weight']=weight
```

```python
print("filled structured array")
print(data)
```

```python

print("all names")
print(data['name'])

```

```python
print("first row of data")
print(data[0])

```

```python
print("names from the last row")
print(data[-1]['name'])
```

```python

print("names where age is under 30")
print(data[data['age'] < 30]['name'])
```

```python

#dictonary method
np.dtype({
    'names':('name','age','weight'),
    'formats':('U10','i4','f8')
})
```

```python

np.dtype({
    'names':('name','age','weight'),
    'formats':((np.str_,10),int,np.float32)
})
```

```python

np.dtype([
    ('name','S10'),
    ('age','i4'),
    ('weight','f8')
])

```

```python

np.dtype('S10,i4,f8')
```

```python
tp=np.dtype([
    ('id','i8'),
    ('mat','f8',(3,3))
])

```

```python
x=np.zeros(1,dtype=tp)
```

```python
print(x[0])
```

```python
print(x['mat'][0])
```

```python
#record array

data_rec=data.view(np.recarray)
print(data_rec.age)
```
