```python
np.random.seed(0)
a= np.random.randint(10, size=10)
print(a)
np.sort(a)
```

```python
np.random.seed(0)
b=np.random.randint(0,50,(4,6))
print(b)
bc=np.sort(b, axis=0)
print(bc)
br=np.sort(b,axis=1)
print(br)
```

```python
np.random.seed(0)
c=np.random.randint(0,50, size=15)
print(c)
ca=np.partition(c,4)
print(ca)
np.sort(c)
```

```python
student_names = np.array(["Anobe", "Bob", "Charle", "David"])
scores = np.array([82, 65, 91, 74])
sort=np.argsort(scores)
for i in sort:
    print(student_names[i])
```

```python
np.random.seed(0)
X = np.random.rand(10, 2)

dist_sq = np.sum((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2, axis=-1)
nearest_neighbors = np.argsort(dist_sq, axis=1)[:, 1:3]

for point_index, neighbors in enumerate(nearest_neighbors):
    print(f"Point {point_index}: {neighbors}")
```

```python
name=['Alice', 'Bob', 'Charlie', 'David']
age=[22,29,30,37]
marks=[85.6, 92.3, 78.9, 88.5]

data=np.zeros(4, dtype=[('name', 'U10'), ('age', 'i4'), ('marks', 'f8')])
data['name'] = name
data['age'] = age
data['marks'] = marks
print(data[data['age'] < 30]['name'])
```

```python
title=['DSA','EDA','ML','AI']
author=['Alice','Bob','Charlie','David']
price=[229.99, 359.99, 429.99, 159.99]

book=np.zeros(4, dtype=[('title', 'U10'), ('author', 'U10'), ('price', 'f8')])
book['title'] = title
book['author'] = author
book['price'] = price
print(book)
```

```python
print(book['title'])
print(book[1])
print(book[book['price'] < 300]['title'])
```
