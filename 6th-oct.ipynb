```python
import numpy as np
import pandas as pd
```

```python
data=pd.Series([1,np.nan,3,None,5],index=['a','b','c','d','e'])
print(data)
```

```python
print(data.isnull())
```

```python
print(data.notnull())
```

```python
data1=pd.Series([1,np.nan,3,None,5],index=['a','b','c','d','e'])
data2=data1.dropna()
print(data1)
print()
print(data2)
```

```python
data3=pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,5]])
print(data3)
print(data3.dropna())
print()
print(data3.dropna(axis='columns'))
print()
print(data3.dropna(axis='columns',how='all'))
print()
print(data3.dropna(axis='rows',thresh=3))
```

```python
data4=pd.DataFrame([[1,np.nan,2],[2,3,5],[np.nan,4,5]])
print(data4)
print()
print(data4.fillna(0))
print()
print(data4.ffill())
print()
print(data4.bfill())
```
