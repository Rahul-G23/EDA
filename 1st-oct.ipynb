```python
name=['Alice','Bob','Charles','David','Edward']
age=[24,42,18,68,53]
hight=[165.2,156.9,177.8,142.5,150.0]

dat=np.zeros(5,dtype={'names':('name','age','hight'),'formats':('U10','i4','f8')})
dat['name']=name
dat['age']=age
dat['hight']=hight
print(dat)
darf=pd.DataFrame(dat)
print(darf)
```

```python
ind=pd.Index([2,3,5,7,11])
print(ind)
print(ind.size,ind.shape,ind.ndim,ind.dtype)

ind[3]=40

print(ind[1])
print(ind[::2])
print(ind[1:4:2])
```

```python
indA=pd.Index([1,3,5,7,9])
indB=pd.Index([2,3,5,7,11])

print(indA & indB)
print(indA | indB)
print(indA ^ indB)
print(indA.intersection(indB))
```
